```python
from google.colab import files

uploaded = files.upload()
```

```python
import pandas as pd

filename = list(uploaded.keys())[0]
df = pd.read_excel(filename)

print("Dataset loaded successfully!")
print(df.head())
```

```python
print("Dataset Shape:", df.shape)
print("Column Names:", df.columns.tolist())
```

```python
total_sales = df["Sales_Amount"].sum()

print("Total Sales:", round(total_sales, 2))
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
product_sales = df.groupby("Product")["Sales_Amount"].sum()

print(product_sales.sort_values(ascending=False))
```

```python
category_sales = df.groupby("Category")["Sales_Amount"].sum()

print(category_sales.sort_values(ascending=False))
```

```python
region_sales = df.groupby("Region")["Sales_Amount"].sum()

print(region_sales.sort_values(ascending=False))
```

```python
customer_sales = df.groupby("Customer_Type")["Sales_Amount"].sum()

print(customer_sales.sort_values(ascending=False))
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")

df["Month"] = df["Order_Date"].dt.to_period("M")

monthly_sales = df.groupby("Month")["Sales_Amount"].sum()

print(monthly_sales)
```

```python
top_products = df.groupby("Product")["Sales_Amount"].sum()

print(top_products.sort_values(ascending=False).head(5))
```

```python
with pd.ExcelWriter("Task_2_Sales_Analysis.xlsx") as writer:
    product_sales.to_excel(writer, sheet_name="Product Sales")
    category_sales.to_excel(writer, sheet_name="Category Sales")
    region_sales.to_excel(writer, sheet_name="Region Sales")
    customer_sales.to_excel(writer, sheet_name="Customer Sales")
    monthly_sales.to_excel(writer, sheet_name="Monthly Sales")

print("Task 2 analysis saved successfully!")
```
